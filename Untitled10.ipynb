```python
# 1 Which rating category has the highest average book price?
import pandas as pd


df = pd.read_csv("scraped_books_with_rating.csv")


df["Price_Clean"] = df["Price"].str.extract(r"([\d\.]+)").astype(float)


avg_df = df.groupby("Rating")["Price_Clean"].mean().reset_index()
highest_rating = avg_df.sort_values(by="Price_Clean", ascending=False).iloc[0]


print(
    f"Rating Category: {highest_rating['Rating']}\nAverage Price: £{highest_rating['Price_Clean']:.2f}"
)
```

```python
3.# •  What percentage of books are available in stock?
import pandas as pd


df = pd.read_csv("scraped_books_with_rating.csv")

total_books = len(df)
in_stock_books = len(df[df["Availability"] == "In stock"])
percentage_in_stock = (in_stock_books / total_books) * 100


print(f"Percentage of Books in Stock: {percentage_in_stock:.2f}%")
```

```python
4.# Does book availability differ depending on the rating?
import pandas as pd


df = pd.read_csv("scraped_books_with_rating.csv")

df['Availability'] = df['Availability'].str.strip()

availability_by_rating = df.groupby('Rating')['Availability'].value_counts(normalize=True) * 100

print(availability_by_rating)

```

```python
5.# Which rating is the most common among the 1,000 books?
import pandas as pd


df = pd.read_csv("scraped_books_with_rating.csv")

ratings_count = df['Rating'].value_counts()

most_common_rating = ratings_count.idxmax()

print(f"The most common rating among the 1,000 books is: {most_common_rating}")
```

```python
import pandas as pd
import scipy.stats as stats
import seaborn as sns
import matplotlib.pyplot as plt

# Load & Clean Data
df = pd.read_csv("scraped_books_with_rating.csv")
df["Price_Clean"] = df["Price"].str.extract(r"([\d\.]+)").astype(float)
rating_map = {"One": 1, "Two": 2, "Three": 3, "Four": 4, "Five": 5}
df["Rating_Num"] = df["Rating"].map(rating_map)

# 1. Trend Analysis: Price Boxplot across Ratings
sns.boxplot(x="Rating_Num", y="Price_Clean", data=df)
plt.title("Price Distribution Across Rating Levels")
plt.show()

# 2. Hypothesis Testing: One-Way ANOVA
groups = [group["Price_Clean"].values for _, group in df.groupby("Rating_Num")]
f_stat, p_val = stats.f_oneway(*groups)

print(f"F-Statistic: {f_stat:.4f}")
print(f"p-value: {p_val:.4f}")

if p_val < 0.05:
    print("Conclusion: Significant price difference exists across ratings.")
else:
    print("Conclusion: No significant price difference found across ratings.")
```
